```python
import os
import shutil

# Define region of interest
roi = "rivermouth_saltmarsh"

# Define full input folder path (relative or absolute)
input_folder = os.path.abspath(f"./Data/{roi}/local_images/PSScene4Band")

# Validate that the folder exists and contains .tif files
if not os.path.isdir(input_folder):
    raise FileNotFoundError(f"Input folder not found: {input_folder}")

tif_files = [f for f in os.listdir(input_folder) if f.lower().endswith('.tif')]
if not tif_files:
    raise ValueError(f"No .tif files found in: {input_folder}")

# Define zip filename
zip_basename = f"{roi}.pss"
zip_filename = f"{zip_basename}.zip"
zip_output_path = os.path.abspath(zip_filename)

# Zip the entire folder (with folder name in archive)
shutil.make_archive(
    base_name=zip_basename,
    format='zip',
    root_dir=os.path.dirname(input_folder),
    base_dir=os.path.basename(input_folder)
)

print("✅ Zip archive created successfully.")
print("📦 Path:", zip_output_path)
print("🗂  Included files:", tif_files)
```
